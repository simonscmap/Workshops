## Colocalize
sample the target datasets at the loacations specified by the source dataset. Here, the source dataset is a cruise trajectory. 

```python

targets = {
            "tblAltimetry_REP_Signal": {
                                    "variables": ["sla", "adt"],
                                    "tolerances": [0, 0.25, 0.25, 5]
                                    },
            "tblDarwin_Nutrient_Climatology": {
                                    "variables": ["PO4_darwin_clim"],
                                    "tolerances": [0, 0.25, 0.25, 5]
                                    },       
            }

src_df = api.cruise_trajectory("Gradients_1")[:3]  # just a few rows for demonstration

pycmap.Sample(
              source=src_df, 
              targets=targets,
              replaceWithMonthlyClimatolog=True,
              agg_fun=["avg", "count", "stdev"]
             )
```
